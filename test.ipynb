```python
import sqlite3

import pandas as pd


def connect_to_db(path):
    """Create a connection to the SQLite database specified by the db_file path"""
    print(f"Connecting to database at {path}")
    conn = sqlite3.connect(path)
    return conn


# Paths to databases
modelset_dir = "data/modelset/"  # path to the modelset dataset which can be downloaded from https://github.com/modelset/modelset-dataset/releases

uml_datasets = ["genmymodel", "ecore"]  # List of UML datasets to process

for uml_dataset in uml_datasets:
    uml_db_path = f"{modelset_dir}/datasets/dataset.{uml_dataset}/data/{uml_dataset}.db"
    analysis_db_path = f"{modelset_dir}/datasets/dataset.{uml_dataset}/data/analysis.db"

    uml_conn = connect_to_db(uml_db_path)
    analysis_conn = connect_to_db(analysis_db_path)

    uml_query_models = "SELECT * FROM models;"
    uml_df_models = pd.read_sql_query(uml_query_models, uml_conn)

    uml_query_metadata = "SELECT * FROM metadata;"
    uml_df_metadata = pd.read_sql_query(uml_query_metadata, uml_conn)

    analysis_query_models = "SELECT * FROM models;"
    analysis_df_models = pd.read_sql_query(analysis_query_models, analysis_conn)

    analysis_query_stats = "SELECT * FROM stats;"
    analysis_df_stats = pd.read_sql_query(analysis_query_stats, analysis_conn)

    print(uml_df_metadata.columns)
    print(uml_df_models.columns)
    print(analysis_df_models.columns)
    print(analysis_df_stats.columns)
```

```python
from mcp4cm import DatasetType, load_dataset

# Load all ArchiMate models, regardless of natural language
all_archimate = load_dataset(
    DatasetType.EAMODELSET,
    "data/eamodelset",
)

# Load only English ArchiMate models
english_archimate = load_dataset(
    DatasetType.EAMODELSET,
    "data/eamodelset",
    language="en",
)

print("All ArchiMate:", len(all_archimate))
print("English ArchiMate:", len(english_archimate))
```

```python
# Load all UML models
uml = load_dataset(DatasetType.MODELSET_UML, "data/modelset")

# Load all Ecore models
ecore = load_dataset(DatasetType.MODELSET_ECORE, "data/modelset")
print("UML models:", len(uml))
print("Ecore models:", len(ecore))
```
